# GM12878 validation

89 CRISPR-tested enhancer-gene pairs in GM12878 (Nasser et al. 2021), 32 positive.
As at GWAS loci, every protein-coding transcript of the gene has its own row, and
each pair is scored by its best transcript. Missing Hi-C contacts count as 0, as in
training and in the preprocessing workflow. The matrix is built by
`EPFinder_training/scripts/build_features.py --transcripts all`.

```python
gm12878 = pd.read_csv("./dataset/GM12878_features_all_transcripts.tsv", sep="\t")
```

```python
gm12878
```

```python
gm12878_29features = gm12878.drop(['#Class', 'ID', 'Enh', 'Prom', 'TX'], axis=1)
gm12878_29features['HiC_Contact'] = gm12878_29features['HiC_Contact'].fillna(0)
```

```python
gm12878_29features
```

```python
gm12878_predict = predict_model(EPFinder_model, raw_score = True, data = gm12878_29features)
```

```python
gm12878_predict
```

```python
gm12878['EPFinder_Prediction'] = gm12878_predict['prediction_score_1'].values

# One score per enhancer-gene pair: its best transcript
pairs = gm12878.groupby('ID', sort=False).agg(
    label=('#Class', 'first'), score=('EPFinder_Prediction', 'max'), transcripts=('TX', 'size'))

auroc = roc_auc_score(pairs['label'], pairs['score'])
auprc = average_precision_score(pairs['label'], pairs['score'])

print(f"{len(pairs)} pairs ({int(pairs['label'].sum())} positive), {len(gm12878)} transcript rows")
print(f"AUROC (Area Under ROC Curve):           {auroc:.4f}")
print(f"AUPRC (Area Under Precision-Recall):    {auprc:.4f}")
```
